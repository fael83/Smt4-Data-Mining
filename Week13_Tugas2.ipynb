```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_blobs, load_iris
```

```python
# Membuat data contoh
X, _ = make_blobs(n_samples=300, centers=4, cluster_std=0.60, random_state=0)
```

```python
# Menggunakan DBSCAN
dbscan = DBSCAN(eps=0.3, min_samples=10)
labels = dbscan.fit_predict(X)
```

```python
# Visualisasi hasil clustering
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='viridis')
plt.title("Hasil DBSCAN Clustering")
plt.show()
```

```python
from sklearn.cluster import AgglomerativeClustering
# Menggunakan Agglomerative Clustering
agg_clust = AgglomerativeClustering(n_clusters=4)
agg_labels = agg_clust.fit_predict(X)
```

```python
# Visualisasi hasil clustering
plt.scatter(X[:, 0], X[:, 1], c=agg_labels,
cmap='viridis')
plt.title("Hasil Agglomerative Clustering")
plt.show()
```

```python
# ubah parameter eps dan min_samples
eps_best = 0.6
min_samples_best = 10

dbscan_optimized = DBSCAN(eps=eps_best, min_samples=min_samples_best)
labels_optimized = dbscan_optimized.fit_predict(X)
```

```python
# Visualisasi Hasil
plt.figure(figsize=(8, 6))
plt.scatter(X[:, 0], X[:, 1], c=labels_optimized, cmap='turbo', s=50)
plt.title(f"Hasil DBSCAN Setelah Tuning (eps={eps_best}, min_samples={min_samples_best})")
plt.grid(True)
plt.show()
```
